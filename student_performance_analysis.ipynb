#Initial finding:Distributin of Final Grades
The final grade G3 are mainly concentrated between 8 and 15 with grade 11 being the most frequently recorded final grade.
Very high grades are uncommon, with only one studentb having a recorded grade of 20.

```python
df.groupby("studytime")["G3"].mean()
```

```python
studytime_avg = df.groupby("studytime")["G3"].mean()
plt.figure(figsize=(8, 5))
plt.bar(studytime_avg.index, studytime_avg.values)
plt.title("Average Final Grade By Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade")
plt.xticks([1,2,3,4], ["<2 hours", "2-5 hours", "5-10 hours", ">10 hours"])
plt.show()
```

#Finding:  time Study time and Final Grade
The analysis suggests an association between study time and final grade. Students who study 5 to 10 hours per weeek had the highest average final grade of 11.40, while students who studied less than 2 hours had the lowest average of 10.05.
However the average grade decsreased slightly to 11.26 among students who studied more than 10 hours.
Therefore, the relationship is not strictly increasing, and we cannot conclude thst studying more hours causes higher grades.

```python
df["failures"].value_counts().sort_index()
```

```python
failure_avg = df.groupby("failures")["G3"].mean()
failure_avg
```

#Students with more previous failures tend to have more lower final grades. Students with no previous failures had the highest average final grade, while students with three previous failures had the lowest.
```python
plt.figure(figsize=(8,5))
plt.bar(failure_avg.index, failure_avg.values)
plt.title("Average Final Grade by Previous Failures")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Average Final Grade")
plt.show()
           
```

# There's a negative association between previous failures and final grade.Students with more previous failures tend to have lower average final grade , while students with no previous failures tend to have a higher average final grades.
However, this association does not prove that previous failures cause lower final grades.
